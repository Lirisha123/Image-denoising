```python
import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import ModelCheckpoint
from google.colab import drive

# ==========================================
# 1. MOUNT GOOGLE DRIVE & SET PATHS
# ==========================================
drive.mount('/content/drive')

save_dir = '/content/drive/MyDrive/PBL_Image_Denoising'
os.makedirs(save_dir, exist_ok=True)

model_path = os.path.join(save_dir, 'autoencoder_fashion.h5')
plot_path = os.path.join(save_dir, 'fashion_denoising_results.png')

# ==========================================
# 2. INSTANT LOAD & PREPROCESS DATASET
# ==========================================
print("[INFO] Loading Fashion-MNIST dataset...")
(x_train, _), (x_test, _) = tf.keras.datasets.fashion_mnist.load_data()

# Normalize pixel values to [0.0, 1.0] and reshape for CNN (28, 28, 1)
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

x_train = np.reshape(x_train, (len(x_train), 28, 28, 1))
x_test = np.reshape(x_test, (len(x_test), 28, 28, 1))

# ==========================================
# 3. ADD SYNTHETIC GAUSSIAN NOISE
# ==========================================
noise_factor = 0.3
x_train_noisy = x_train + noise_factor * np.random.normal(loc=0.0, scale=1.0, size=x_train.shape)
x_test_noisy = x_test + noise_factor * np.random.normal(loc=0.0, scale=1.0, size=x_test.shape)

x_train_noisy = np.clip(x_train_noisy, 0., 1.).astype('float32')
x_test_noisy = np.clip(x_test_noisy, 0., 1.).astype('float32')

# ==========================================
# 4. BUILD CONVOLUTIONAL AUTOENCODER (CAE)
# ==========================================
input_img = layers.Input(shape=(28, 28, 1))

# --- ENCODER ---
x = layers.Conv2D(32, (3, 3), activation='relu', padding='same')(input_img)
x = layers.MaxPooling2D((2, 2), padding='same')(x)
x = layers.Conv2D(16, (3, 3), activation='relu', padding='same')(x)
encoded = layers.MaxPooling2D((2, 2), padding='same')(x)

# --- DECODER ---
x = layers.Conv2D(16, (3, 3), activation='relu', padding='same')(encoded)
x = layers.UpSampling2D((2, 2))(x)
x = layers.Conv2D(32, (3, 3), activation='relu', padding='same')(x)
x = layers.UpSampling2D((2, 2))(x)

# Output layer: 1 filter for grayscale with sigmoid activation
decoded = layers.Conv2D(1, (3, 3), activation='sigmoid', padding='same')(x)

autoencoder = models.Model(input_img, decoded)
autoencoder.compile(optimizer='adam', loss='mean_squared_error')

# ==========================================
# 5. TRAIN & SAVE TO GOOGLE DRIVE
# ==========================================
checkpoint = ModelCheckpoint(
    filepath=model_path,
    monitor='val_loss',
    save_best_only=True,
    verbose=1
)

print("\n[INFO] Starting model training...")
history = autoencoder.fit(
    x_train_noisy, x_train,
    epochs=10,
    batch_size=128,
    shuffle=True,
    validation_data=(x_test_noisy, x_test),
    callbacks=[checkpoint]
)

# ==========================================
# 6. EVALUATE METRICS & SAVE PLOT
# ==========================================
decoded_imgs = autoencoder.predict(x_test_noisy)

x_test_tf = tf.cast(x_test, tf.float32)
x_test_noisy_tf = tf.cast(x_test_noisy, tf.float32)
decoded_imgs_tf = tf.cast(decoded_imgs, tf.float32)

psnr_noisy = tf.image.psnr(x_test_tf, x_test_noisy_tf, max_val=1.0)
psnr_denoised = tf.image.psnr(x_test_tf, decoded_imgs_tf, max_val=1.0)

ssim_noisy = tf.image.ssim(x_test_tf, x_test_noisy_tf, max_val=1.0)
ssim_denoised = tf.image.ssim(x_test_tf, decoded_imgs_tf, max_val=1.0)

print("\n==========================================")
print("       QUANTITATIVE EVALUATION RESULTS    ")
print("==========================================")
print(f"PSNR (Before Denoising): {np.mean(psnr_noisy):.2f} dB")
print(f"PSNR (After Denoising) : {np.mean(psnr_denoised):.2f} dB")
print(f"SSIM (Before Denoising): {np.mean(ssim_noisy):.4f}")
print(f"SSIM (After Denoising) : {np.mean(ssim_denoised):.4f}")
print("==========================================")

# Plot & Save Visualization
n = 5
plt.figure(figsize=(12, 6))
for i in range(n):
    # Noisy Input
    plt.subplot(3, n, i + 1)
    plt.imshow(x_test_noisy[i].reshape(28, 28), cmap='gray')
    plt.title("Noisy Input")
    plt.axis('off')

    # Denoised Output
    plt.subplot(3, n, i + 1 + n)
    plt.imshow(decoded_imgs[i].reshape(28, 28), cmap='gray')
    plt.title("Denoised Output")
    plt.axis('off')

    # Ground Truth
    plt.subplot(3, n, i + 1 + 2 * n)
    plt.imshow(x_test[i].reshape(28, 28), cmap='gray')
    plt.title("Ground Truth")
    plt.axis('off')

plt.tight_layout()
plt.savefig(plot_path)
plt.show()

print(f"\n[SUCCESS] Model saved to Drive at:\n{model_path}")
print(f"[SUCCESS] Visual plot saved to Drive at:\n{plot_path}")
```
